```python
import numpy as np
import matplotlib.pyplot as plt
from skimage import data
# from skimage import data gives us access to built-in sample images.
print("all libraries imported successfully!")
```

```python
#load a built-in sample photograph
img = data.astronaut()

#display information about the image
print("Image shape:", img.shape)
print("Data types:", img.dtype)

#display the original image
plt.figure(figsize=(10,10))
plt.title("Astronaut")
plt.imshow(img)
plt.show()
```

```python
#extract the three color channels
red = img[:, :, 0]
green = img[:, :, 1]
blue = img[:, :, 2]

#calculate grayscale brightness
gray = (
    0.299 * red +
    0.587 * green +
    0.11 * blue
).astype(np.uint8)

#display the grayscale image
plt.figure(figsize=(10,10))
plt.title("Grayscale")
plt.imshow(gray, cmap="gray")
plt.show()
```

```python

# Flip from left to right
horizontal_flip = np.fliplr(img)

# Flip from top to bottom
vertical_flip = np.flipud(img)

# Display both images
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(horizontal_flip)
plt.title("Horizontal Flip")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(vertical_flip)
plt.title("Vertical Flip")
plt.axis("off")

plt.show()
```

```python
from numpy._core.multiarray import RAISE
# rotate 90 degrees clockwise
rotated_90 = np.rot90(img)

plt.figure(figsize=(10,10))
plt.title("Rotated 90 degrees")
plt.imshow(rotated_90)
plt.show()

print("Orignial shape:", img.shape)
print("Rotated shape:", rotated_90)
```

```python

# Convert grayscale image into a 3-channel image
gray_rgb = np.stack([gray, gray, gray], axis=2)

print("Grayscale RGB shape:", gray_rgb.shape)
```

```python
#extract the RGB channels
red_channel = img[:, :, 0]
green_channel = img[:, :, 1]
blue_channel = img[:, :, 2]

#display each channel
plt.figure(figsize=(12,4))
plt.subplot(1, 3, 1)
plt.title("Red Channel")
plt.imshow(red_channel, cmap="gray")

plt.subplot(1, 3, 2)
plt.title("Green Channel")
plt.imshow(green_channel, cmap="gray")

plt.subplot(1, 3, 3)
plt.title("Blue Channel")
plt.imshow(blue_channel)

print("Red channel:", red_channel.shape)
print("Green channel:", green_channel.shape)
print("Blue channel:", blue_channel.shape)
```

```python

# Step 1: Convert grayscale into 3 channels
gray_rgb = np.stack([gray, gray, gray], axis=2)

# Step 2: Create flipped and rotated images
horizontal_flip = np.fliplr(img)
rotated_img = np.rot90(img, k=-1)

# Step 3: Combine images horizontally
top_row = np.concatenate((img, gray_rgb), axis=1)
bottom_row = np.concatenate(
    (horizontal_flip, rotated_img), axis=1
)

# Step 4: Combine both rows vertically
collage = np.concatenate((top_row, bottom_row), axis=0)

# Step 5: Display the collage
plt.figure(figsize=(10, 10))
plt.imshow(collage)
plt.title("NumPy Image Processing Toolkit")
plt.axis("off")
plt.show()

print("Collage shape:", collage.shape)
print("Collage created successfully!")
```
