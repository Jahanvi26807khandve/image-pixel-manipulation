```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = np.zeros((200, 200, 3), dtype=np.uint8)

print("Image shape:", img.shape)
print("Data type:", img.dtype)
print("Memory usage:", img.nbytes, "bytes")

plt.imshow(img)
plt.title("Blank RGB Image")
plt.axis("off")
plt.show()
```

```python
img = np.zeros((200, 200, 3), dtype=np.uint8)   # black image (height, width, RGB)

# fill a region with a solid color using np.full
img[0:100, 0:100] = np.full((100, 100, 3), [255, 0, 0], dtype=np.uint8)   # red block

plt.imshow(img)
plt.title("Blank image with red block")
plt.show()
```

```python

# Draw colored blocks
img[10:60, 10:60] = [255, 0, 0]       # Red block
img[10:60, 70:120] = [0, 255, 0]     # Green block
img[10:60, 130:180] = [0, 0, 255]    # Blue block

# Draw horizontal stripes
img[80:100, :] = [255, 255, 0]       # Yellow
img[110:130, :] = [255, 0, 255]      # Magenta

# Display the image
plt.figure(figsize=(6, 6))
plt.imshow(img)
plt.title("Synthetic RGB Image")
plt.axis("off")
plt.show()
```

```python
# green block: rows 10-60, cols 20-90 -> set channel 1 (G) to 255
img[10:60, 20:90, 1] = 255

# blue block
img[120:180, 20:90] = [0, 0, 255]

# horizontal stripes (every 20 rows, 5 pixels thick)
for r in range(0, 200, 20):
    img[r:r+5, 110:190] = [255, 255, 0]   # yellow

# vertical stripe
img[:, 100:105] = [255, 255, 255]          # white

plt.imshow(img)
plt.title("Blocks and stripes")
plt.show()
```

```python
np.random.seed(42)
img = np.random.randint(0, 256, size=(200, 200, 3), dtype=np.uint8)

plt.imshow(img)
plt.title("Random colours")
plt.axis("off")
plt.show()
```
